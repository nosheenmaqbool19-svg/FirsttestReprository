# 3-Dictionaries

-An unordered collection of elements

-key and values

-Curly braces or barckets{}

-muteable/change the values


```python
# -food and their prices
food1 = {"samosa":30, "pakora":100, "Raita":20, "Salad":50, "chicken rolls":30}
food1
```

```python
type(food1)
```

```python
# extract data
keys1 = food1.keys()
keys1
```

```python
value1 = food1.values()
value1
```

```python
# adding new element
food1["Tikki"]= 10
food1
```

```python
# update the value
food1["Tikki"] =15
food1
```

```python
food2 ={"Dates":50, "Chocolates":100, "Sawiyan":1000}
food2
```

```python
# -concatinate
food1.update(food2)
food1
```
