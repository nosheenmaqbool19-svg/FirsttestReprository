```python
          
#1-import liberaries
import seaborn as sns
import matplotlib.pyplot as plt

#2-set aesthetic background
sns.set_theme(style="darkgrid")

#3-load dataset
flights = sns.load_dataset("flights")

# 4-draw bar plot using mean as estimator
# Note: estimator=mean is the default, so we can omit it, or pass the function from numpy if explicitly changing it.
sns.barplot(x="year", y="passengers", data=flights, palette="colorblind")

# 5-Add custom titles and labels (Must come after drawing the plot)
plt.title("Average Flight Passengers per Year", fontsize=14, fontweight="bold")
plt.xlabel("year")
plt.ylabel("Average Number of Passengers")
# 6-display the plot
plt.show()
```

```python
# 1-import libraries
import matplotlib.pyplot as plt
import seaborn as sns

# 2-set aesthetic background
sns.set_theme(style="darkgrid")

# 3-load dataset
flights = sns.load_dataset("flights")

# 4-draw bar plot using mean as estimator
sns.barplot(x="year", y="passengers", data=flights, palette="deep")

# 5-Add custom titles and labels (Removed the '=' signs)
plt.title("Average Flight Passengers per Year", fontsize=14, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Average Number of Passengers")

# 6-display the plot
plt.show()
```
