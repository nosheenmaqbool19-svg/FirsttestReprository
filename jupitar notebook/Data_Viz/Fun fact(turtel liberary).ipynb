# Heart shape

```python
import turtle
screen = turtle.Screen()
screen.title("Heart Drawing")

t = turtle.Turtle()
t.speed(3)
t.color("red")

t.begin_fill()
t.left(140)
t.forward(113)
t.circle(-57, 200)
t.left(120)
t.circle(-57, 200)
t.forward(113)
t.end_fill()

screen.update()
turtle.done()
```

```python
from ipyturtle3 import Turtle, Canvas

my_canvas = Canvas(width=400, height=400)
display(my_canvas)

t = Turtle(my_canvas)
t.shape("turtle")
t.color("red")

t.begin_fill()
t.left(140)
t.forward(113)
t.circle(-57, 200)
t.left(120)
t.circle(-57, 200)
t.forward(113)
t.end_fill()
```

```python
from ipyturtle3 import Turtle, Canvas, TurtleScreen

my_canvas = Canvas(width=400, height=400)
display(my_canvas)

# Wrap the canvas in a TurtleScreen first
my_screen = TurtleScreen(my_canvas)
my_screen.clear()

# Pass my_screen (not my_canvas) to Turtle
t = Turtle(my_screen)
t.shape("turtle")
t.color("red")

t.begin_fill()
t.left(140)
t.forward(113)
t.circle(-57, 200)
t.left(120)
t.circle(-57, 200)
t.forward(113)
t.end_fill()
```
