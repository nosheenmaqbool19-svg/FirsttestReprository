## Code Example: Box Plot with the "Tips" Dataset
Let's see how much restaurant bills vary depending on the day of the week:

```python
#1-import liberaries
import seaborn as sns
import matplotlib.pyplot as plt

 #-2 set aesthetic background
sns.set_theme(style="whitegrid")
 #-3 load dataset
tips = sns.load_dataset("tips")
#-4 draw boxplot
# This breaks down the 'total_bill' distribution for each 'day'
sns.boxplot(x="day", y="total_bill", data=tips, hue="day", palette="pastel")
#-5 add professional titles and custom labels
plt.title("Distribution of Total Bills by Day of the Week", fontsize=14, fontweight="bold")
plt.xlabel("Day")
plt.ylabel("Total Bill Amount ($)")
plt.show()
```

```python
#1-import liberaries
import seaborn as sns

 #-2 set aesthetic background
sns.set_theme(style="whitegrid")
 #-3 load dataset
tip = sns.load_dataset("tips")

# 4 - Define individual colors for each 'smoker' hue
smoker_colors = {"Yes":"#32a83e", #Green  
                "No":"#a83242" #red 
}
#-4 draw boxplot
# This breaks down the 'total_bill' distribution for each 'day'
sns.boxplot(x ="tip", y="day", hue= "smoker", data=tip ,dodge=True, orient='h', palette = smoker_colors)
```

```python
# 1 - Import libraries
import seaborn as sns
import matplotlib.pyplot as plt

# 2 - Set aesthetic background
sns.set_theme(style="whitegrid")

# 3 - Load dataset
tip = sns.load_dataset("tips")

# 4 - Define individual colors for each 'smoker' hue
smoker_colors = {"Yes": "#e63946",   # Red (or any hex code / color name)
    "No": "#457b9d"     # Blue
}

# 5 - Draw boxplot
sns.boxplot(x="tip", y="day", hue="smoker", data=tip, palette=smoker_colors, dodge=True, orient="h")

plt.show()
```
