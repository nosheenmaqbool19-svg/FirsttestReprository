# Plotly 

```python
import plotly.express as px

# Load the built-in restaurant tips dataset
df = px.data.tips()
# Generate an interactive scatter plot in one line
fig = px.scatter(df, x="total_bill", y="tip", color="smoker", title="Restaurant Tips Analysis")
fig.show()
```

```python
import plotly.express as px
df = px.data.iris()
fig =px.scatter(df, x="sepal_width", y="sepal_length", color="species", marginal_y="violin", marginal_x="box", trendline="ols", template="simple_white")
fig.show()
```

```python

#Task Requirements

• Dataset: Load the built-in restaurant tips dataset using df = px.data.tips().
• Main Plot: Create a scatter plot mapping total_bill on the x-axis and tip on the y-axis.
• Color Grouping: Color code the data points by whether the diner is a smoker or non-smoker (color="smoker").
• Subplots (Marginal Layouts): Add a histogram on the marginal x-axis and a box plot on the marginal y-axis.
• Analysis Line: Add an OLS trendline to see the rate of increase for each group.
• Sub-charts (Faceting): Split the visualization into side-by-side subplots based on the dining time of day (facet_col="time").
• Theme: Apply the "plotly_dark" layout template for a modern aesthetic.
```

```python
import plotly.express as px
df = px.data.tips()
fig =px.scatter(df, x="total_bill", y="tip", color="smoker", marginal_x="histogram", marginal_y="box plot", trendline="ols",facet_col="time", template="plotly_dark") 
fig.show()
```

```python
import plotly.express as px
df = px.data.gapminder()
fig =px.scatter(df, x="gdpPercap", y="lifeExp", animation_frame="year", animation_group="country", size="pop", color="continent", hover_name="country", facet_col="continent", log_x="True", size_max=45, range_x=[100, 100000], range_y=[25, 90])
fig.show()
```

```python
import plotly.express as px
df = px.data.gapminder().query("year== 2007").query("continent=='Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'other countries' #Represent only large countries
fig =px.pie(df, values="pop", names='country', title= 'Population of European continents')
fig.show()
```

```python
import plotly.express as px
df = px.data.gapminder().query("year== 2007").query("continent=='Europe'")
fig =px.sunburst(df, path=['continent','country'], values="pop",color='lifeExp', hover_data=['iso_alpha'])
fig.show()
```
