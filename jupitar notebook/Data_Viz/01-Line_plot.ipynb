## Code Example: From Bar to Line Plot


```python
# 1-import liberaries
import seaborn as sns
import matplotlib.pyplot as plt

# 2-set an aesthetic background
sns.set_theme(style="darkgrid")

 # 3-load dataset
flights = sns.load_dataset("flights")
# 4-draw line plot(change the function name)
sns.lineplot(x="year", y="passengers", data=flights, color="purple", linewidth=2)

#add custom titles and labels
plt.title("Flight Passengers Growth Trend (1949-1960", fontsize=14, fontweight="bold")
plt.xlabel("year")
plt.ylabel("passengers")
plt.show()       
          
```

##challenge Task
Write a script to create a line plot that meets these requirements:
1. Load the "flights" dataset.
2. Put "year" on the X-axis.
3. Put "passengers" on the Y-axis.
4. Separate the trend lines by "month" (hint: use the hue parameter).
5. Set the theme style to "ticks".
6. Add a Bold Title: "Monthly Passenger Growth Trends".
7. Label the X-axis: "Year".
8. Label the Y-axis: "Total Passengers".

```python
#-1 import liberaries
import seaborn as sns
import matplotlib.pyplot as plt

 #-2 set aesthetic background
sns.set_theme(style="ticks")
# -3 load dataset
flights = sns.load_dataset("flights")
#-4 draw lineplot
sns.lineplot(x="year", y="passengers", data=flights, hue="month", color="bright", linewidth=2)
#-5 add title and custom label
plt.title("Monthly Passenger Growth Trends", fontsize=14, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Total Passengers")
plt.show()
```

```python
# -1 import libraries
import matplotlib.pyplot as plt
import seaborn as sns

# -2 set aesthetic background
sns.set_theme(style="ticks")

# -3 load dataset
flights = sns.load_dataset("flights")

# -4 draw lineplot (Fixed hue quotes, changed color to palette)
sns.lineplot(
    x="year",
    y="passengers",
    data=flights,
    hue="month",
    palette="deep",
    linewidth=1.5,
)

# -5 add title and custom label (Changed fontweight to fontstyle)
plt.title("Monthly Passenger Growth Trends", fontsize=14, fontstyle="italic")
plt.xlabel("Year")
plt.ylabel("Total Passengers")

# -6 display the plot
plt.show()
```

## Code Example: Box Plot with the "Tips" Dataset
Let's see how much restaurant bills vary depending on the day of the week:


```python
#1-import liberaries
import seaborn as sns
import matplotlib.pyplot as plt

 #-2 set aesthetic background
sns.set_theme(style="whitegrid")
 #-3 load dataset
tips = sns.load_dataset("tips")
#-4 draw boxplot
# This breaks down the 'total_bill' distribution for each 'day'
sns.boxplot(x="day", y="total_bill", data=tips, hue="day", palette="pastel")
#-5 add professional titles and custom labels
plt.title("Distribution of Total Bills by Day of the Week", fontsize=14, fontweight="bold")
plt.xlabel("Day")
plt.ylabel("Total Bill Amount ($)")
plt.show()
```
